```python
bronze_users_df = spark.table("workspace.default.bronze_users")
```

```python
bronze_users_df.show()
```

```python
bronze_users_df.printSchema()
```

```python
bronze_users_df.count()
```

```python
silver_users_df = bronze_users_df.dropDuplicates()
```

```python
silver_users_df = silver_users_df.dropna()
```

```python
silver_users_df.show()
```

```python
silver_users_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.silver_users")
```
