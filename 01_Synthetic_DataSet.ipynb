```python
import random
import pandas as pd

first_names = [ "Sahib", "Nakul", "Anisha", "Binil", "Harmanjit",
    "Aman", "Karan", "Simran", "Priya", "Rahul"]
last_names = ["Singh", "Yadav", "Chaudhary", "John", "Patel",
    "Sharma", "Kaur", "Brown", "Wilson", "Taylor"]

departments = [ "IT",
    "Finance",
    "HR",
    "Security",
    "Sales",
    "Marketing",
    "Cloud",
    "Engineering"]

countries = ["Canada", "USA", "India", "UK"]

account_statuses = ["Active", "Inactive"]

users = []

for i in range(100):

    user_id = i + 1

    first_name = random.choice(first_names)
    last_name = random.choice(last_names)

    full_name = first_name + " " + last_name

    department = random.choice(departments)

    country = random.choices(
        countries,
        weights=[75,10,10,5]
    )[0]

    account_status = random.choices(
        account_statuses,
        weights=[90,10]
    )[0]

    users.append([user_id, full_name, department, country, account_status])

df = pd.DataFrame(users, columns=["UserID", "UserName", "Department", "Country", "AccountStatus"])
print(df)
```

```python
devices = []

device_types = ["Laptop", "Desktop", "Mobile", "Tablet"]
operating_systems = ["Windows", "macOS", "Linux", "Android", "iOS"]
trust_devices = ["Trusted", "Untrusted"]

for i in range(150):

    device_id = i + 1
    user_id = random.randint(1, 100)
    device_type = random.choice(device_types)

    if device_type == "Laptop":
        operating_system = random.choice(["Windows", "macOS", "Linux"])

    elif device_type == "Desktop":
        operating_system = random.choice(["Windows", "macOS", "Linux"])

    elif device_type == "Mobile":
        operating_system = random.choice(["Android", "iOS"])

    else:
        operating_system = random.choice(["Android", "iPadOS"])

    trust_device = random.choices(trust_devices, weights=[90, 10])[0]

    devices.append([
        device_id,
        user_id,
        device_type,
        operating_systems,
        trust_device
    ])

devices_df = pd.DataFrame(
    devices,
    columns=[
        "DeviceID",
        "UserID",
        "DeviceType",
        "OperatingSystem",
        "TrustDevice"
    ]
)

display(devices_df)
```

```python
# login logs dataset 
login_logs = []
from datetime import datetime, timedelta
login_methods =["Password","MFA","SSO"]
login_statuses = ["Success","Failure"]
risk_levels = ["Low","Medium","High"]
login_countries = ["Canada","USA","India","UK"]
start_date = datetime(2026, 1, 1)
for i in range(500):
    login_id = i+1
    selected_device = devices_df.sample(1).iloc[0]
    user_id = selected_device["UserID"]
    device_id = selected_device["DeviceID"]
    random_days = random.randint(0, 180)
    random_hours = random.randint(0, 23)
    random_minutes = random.randint(0, 59)

    login_time = start_date + timedelta(
        days=random_days,
        hours=random_hours,
        minutes=random_minutes
    )

    ip_address = ".".join([
        str(random.randint(1, 255)),
        str(random.randint(0, 255)),
        str(random.randint(0, 255)),
        str(random.randint(1, 255))
    ])

    country = random.choice(login_countries)
    login_method = random.choice(login_methods)
    
    login_status = random.choices(
    login_statuses,
    weights=[90, 10])[0]

    risk_level = random.choices(
    risk_levels,
    weights=[70, 20, 10])[0]

    login_logs.append([
        login_id,
        user_id,
        device_id,
        login_time,
        ip_address,
        country,
        login_method,
        login_status,
        risk_level
    ])

login_logs_df = pd.DataFrame(
    login_logs,
    columns=[
        "LoginID",
        "UserID",
        "DeviceID",
        "LoginTime",
        "IPAddress",
        "Country",
        "LoginMethod",
        "LoginStatus",
        "RiskLevel"
    ]
)

display(login_logs_df)
login_logs_df.to_csv("login_logs.csv", index = False)
os.listdir()
    



    
```

```python
#  4 th data set cyber alert
alerts = []
alert_id = 1
for index, row in login_logs_df.iterrows():
    if row["RiskLevel"] == "High" and row["LoginStatus"] == "Failure":
        alerts.append([
        alert_id,
        row["LoginID"],
        row["UserID"],
        "Brute Force",
        "High"
        ])
    elif row["RiskLevel"] == "High" and row["LoginStatus"] == "Success":
        alerts.append([
        alert_id,
        row["LoginID"],
        row["UserID"],
        "Suspicious Login",
        "High"
    ])   
        alert_id += 1
alerts_df = pd.DataFrame(
    alerts,
    columns=[
        "AlertID",
        "LoginID",
        "UserID",
        "AlertType",  
        "AlertSeverity"
    ]
)
display(alerts_df)
     
```
